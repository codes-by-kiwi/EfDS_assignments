#fixing column name in PL for standardization

```python
PL.rename(columns={'heigth': 'height'}, inplace=True)


```

#fixing unknown or inf values

```python
PL['height'] = PL['height'].fillna(PL['height'].mean())
PL['weight'] = PL['weight'].fillna(PL['weight'].mean())
PL['sex'] = PL['sex'].fillna(PL['sex'].mode()[0])
PL['smokes'] = PL['smokes'].fillna(PL['smokes'].mode()[0])

print(PL)
print(PL.isna().sum().sum())
```

```python
US['height'] = US['height'].fillna(US['height'].mean())

print(US)
print(US.isna().sum().sum())
```

#Checking height, weight units in each dataset and conversion

```python
print(NL[['height', 'weight']].describe())  #height in cm and weight in kg
```

```python
print(PL[['height', 'weight']].describe())  #height in cm and weight in kg
```

```python
print(UK[['height', 'weight']].describe())
#the lowest height of -1.0000 might suggest there is an incorrect data entry


negative_height_UK_checker = UK[UK['height'] < 0]
negative_weight_UK_checker = UK[UK['weight'] < 0]
print("Negative height values in UK dataset:")
print(negative_height_UK_checker)
print("Negative weight values in UK dataset:")
print(negative_weight_UK_checker)

min_height_UK = UK[UK['height'] > 0]['height'].min()
UK['height'] = UK['height'].apply(lambda x: min_height_UK if x < 0 else x)

print(UK[['height', 'weight']].describe())   # height in inches and weight in pounds
```

```python
print(US[['height', 'weight']].describe())  #height in cm and weight in kg
```

```python
#convert inches to cm in UK dataset for height
UK['height']= UK['height']*2.54
print(UK[['height', 'weight']].describe())

UK['weight']= UK['weight']*0.453592
print(UK[['height', 'weight']].describe())
```

#calculate BMI for each dataset

```python
for df in [NL, PL, UK, US]:
    df['BMI'] = df['weight'] / ((df['height'] / 100) ** 2)
```

```python
print(NL[['height', 'weight', 'BMI']].describe())
print(PL[['height', 'weight', 'BMI']].describe())
print(UK[['height', 'weight', 'BMI']].describe())
print(US[['height', 'weight', 'BMI']].describe())
```

#add the population column to each dataset

```python
NL['population'] = 'NL'
PL['population'] = 'PL'
UK['population'] = 'UK'
US['population'] = 'US'
```

```python
print(NL)
```

#merging datasets

```python
anthropometric_concatenated = pd.concat([NL, PL, UK, US], ignore_index=True)
print(anthropometric_concatenated)
```

```python
combined_dataset = pd.merge(anthropometric_concatenated, genes, on='id', how='inner')
print(combined_dataset)
print(combined_dataset.shape)
```

```python
import os

os.makedirs('assignment_c/out', exist_ok=True)
combined_dataset.to_csv('assignment_c/out/combined_dataset.csv', index=False)
```

#plot BMI against each gene

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
graph_palette = sns.color_palette("coolwarm", as_cmap=True)

plt.figure(figsize=(15, 5))

for i, gene in enumerate(['geneA', 'geneB', 'geneC'], 1):
    plt.subplot(1, 3, i)

    sns.regplot(
        data=combined_dataset,
        x='BMI', y=gene,
        scatter_kws={'alpha': 0.6, 's': 40},
        line_kws={"color": "black", "linewidth": 2.0},
        color=sns.color_palette("Set2")[i]
    )

    plt.title(f'BMI vs the {gene}')
    plt.xlabel('BMI')
    plt.ylabel(f'{gene}')

plt.tight_layout(rect=[0, 0, 1, 0.75])
sns.despine()
plt.show()
```
